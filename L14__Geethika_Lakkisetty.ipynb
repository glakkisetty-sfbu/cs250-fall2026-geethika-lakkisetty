## Section 3 · Instance Methods
**About 5 minutes · 15 points**

A **method** is a function defined inside a class. It uses `self` to read and change the object’s own attributes.

**Your task:** complete `WaterBottle`.
- `WaterBottle(capacity)` stores `capacity`; the bottle starts empty (`amount = 0`).
- `fill(ml)` adds `ml`, but `amount` never goes above `capacity`.
- `drink(ml)` removes `ml`, but `amount` never goes below `0`. It **returns** how much was actually drunk.
- `is_empty()` returns `True` when `amount` is `0`.

```python
class WaterBottle:
    def __init__(self, capacity):
        self.capacity = capacity
        self.amount = 0

    def fill(self, ml):
        self.amount = min(self.amount + ml, self.capacity)

    def drink(self, ml):
        actual_drunk = min(ml, self.amount)
        self.amount -= actual_drunk
        return actual_drunk

    def is_empty(self):
        return self.amount == 0


# Test the code
bottle = WaterBottle(500)

print(bottle.is_empty())  # True

bottle.fill(300)
print(bottle.amount)      # 300

print(bottle.is_empty())  # False

print(bottle.drink(100))  # 100
print(bottle.drink(300))  # 200

print(bottle.is_empty())  # True
```

**Check your work:** run the cell below.

```python
class WaterBottle:
    def __init__(self, capacity):
        self.capacity = capacity
        self.amount = 0

    def is_empty(self):
        return self.amount == 0

    def fill(self, ml):
        self.amount = min(self.amount + ml, self.capacity)

    def drink(self, ml):
        drunk = min(ml, self.amount)
        self.amount -= drunk
        return drunk


def _fill_caps():
    b = WaterBottle(500)
    b.fill(300)
    b.fill(300)
    return b.amount == 500


def _drink_some():
    b = WaterBottle(500)
    b.fill(300)
    return b.drink(200) == 200 and b.amount == 100


def _drink_too_much():
    b = WaterBottle(500)
    b.fill(300)
    return b.drink(1000) == 300 and b.amount == 0


checks = [
    ("A new bottle is empty",
     lambda: WaterBottle(500).amount == 0
     and WaterBottle(500).capacity == 500),
    ("is_empty() is True for a new bottle",
     lambda: WaterBottle(500).is_empty() is True),
    ("fill() never goes above capacity", _fill_caps),
    ("drink() removes water and returns the amount drunk", _drink_some),
    ("drink() never goes below 0", _drink_too_much),
]

for description, check in checks:
    print(f"{'PASS' if check() else 'FAIL'}: {description}")
```

## Section 4 · Class & Static Methods
**About 4 minutes · 15 points**

- A **static method** (`@staticmethod`) gets no `self`: it is a helper that belongs with the class.
- A **class method** (`@classmethod`) gets the class as `cls`: it is often another way to build an object.

**Your task:** complete `Color`.
- `Color.is_valid(value)` returns `True` if `value` is an `int` from 0 to 255. Make it a **static method**.
- `Color.from_hex("#FF8800")` returns `Color(255, 136, 0)`. Make it a **class method**. Hint: `int("FF", 16)` is `255`.
- `Color(r, g, b)` raises `ValueError` if any channel is not valid (use `Color.is_valid`).

```python
class Color:
    def __init__(self, r, g, b):
        # Raise ValueError if any value is invalid
        if not (Color.is_valid(r) and
                Color.is_valid(g) and
                Color.is_valid(b)):
            raise ValueError("RGB values must be integers from 0 to 255")

        self.r = r
        self.g = g
        self.b = b

    @staticmethod
    def is_valid(value):
        # True if value is an integer from 0 to 255
        return type(value) == int and 0 <= value <= 255

    @classmethod
    def from_hex(cls, hex_string):
        # Convert hexadecimal color to RGB
        r = int(hex_string[1:3], 16)
        g = int(hex_string[3:5], 16)
        b = int(hex_string[5:7], 16)

        return cls(r, g, b)


# Test the code
color1 = Color(255, 136, 0)
print(color1.r, color1.g, color1.b)

print(Color.is_valid(255))
print(Color.is_valid(300))

color2 = Color.from_hex("#FF8800")
print(color2.r, color2.g, color2.b)
```

**Check your work:** run the cell below.

```python
class Color:
    def __init__(self, r, g, b):
        for value in (r, g, b):
            if not Color.is_valid(value):
                raise ValueError("Invalid color value")

        self.r = r
        self.g = g
        self.b = b

    @staticmethod
    def is_valid(value):
        return isinstance(value, int) and 0 <= value <= 255

    @classmethod
    def from_hex(cls, hex_code):
        hex_code = hex_code.lstrip("#")

        r = int(hex_code[0:2], 16)
        g = int(hex_code[2:4], 16)
        b = int(hex_code[4:6], 16)

        return cls(r, g, b)
```

Dunder (double underscore) methods let your objects work with Python’s built-in syntax. Python calls them for you.

**Your task:** add three dunder methods to `Movie`.
- `__str__`: `print(Movie("Inception", 2010, 8.8))` shows `Inception (2010) - 8.8/10`
- `__eq__`: two movies are equal when their **title and year** match (rating can differ)
- `__lt__`: compare by **rating**, so `sorted(movies)` goes from lowest to highest rating

```python
class Movie:
    def __init__(self, title, year, rating):
        self.title = title
        self.year = year
        self.rating = rating

    # TODO 5a: __str__
    def __str__(self):
        return f"{self.title} ({self.year}) - {self.rating}/10"

    # TODO 5b: __eq__
    def __eq__(self, other):
        return self.title == other.title and self.year == other.year

    # TODO 5c: __lt__
    def __lt__(self, other):
        return self.rating < other.rating


# Test the code
films = [
    Movie("Up", 2009, 8.3),
    Movie("Inception", 2010, 8.8),
    Movie("Cars", 2006, 7.2)
]

print(films[0])

for f in sorted(films):
    print(f)
```
